## How DDPM Differs from NCSN

| Aspect | NCSN (Notebooks 01-04) | DDPM |
|--------|----------------------|------|
| **Training objective** | Learn score $\nabla_x \log p_\sigma(x)$ | Learn noise prediction $\epsilon_\theta(x_t, t)$ |
| **Noise levels** | Discrete geometric schedule ($L$ levels) | Continuous diffusion process ($T$ timesteps) |
| **Forward process** | Add noise at sampled $\sigma$ | Gradual Gaussian noising via $\beta_t$ schedule |
| **Sampling** | Annealed Langevin dynamics | Reverse diffusion process (ancestral sampling) |
| **Connection** | Score matching: $s_\theta \approx \nabla_x \log p_\sigma$ | Equivalent: $\epsilon_\theta$ and score are related by $s = -\epsilon / \sigma$ |
| **Correction** | Modified Langevin adds explicit correction | DDPM/DDIM use variance scheduling for similar effect |
